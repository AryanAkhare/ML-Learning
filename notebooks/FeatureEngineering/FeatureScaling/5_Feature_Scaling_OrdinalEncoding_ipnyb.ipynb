##Dataset1

```python
import pandas as pd

data = {
    "student": ["Aman", "Riya", "Kunal", "Neha", "Arjun"],
    "study_hours": ["Low", "Medium", "High", "Medium", "Low"],
    "parent_support": ["Poor", "Average", "Good", "Good", "Average"],
    "performance": ["Below Avg", "Average", "Excellent", "Good", "Average"]
}

df = pd.DataFrame(data)
df
```

Step 1: Define order

```python
from sklearn.preprocessing import OrdinalEncoder

study_order = ["Low", "Medium", "High"]
support_order = ["Poor", "Average", "Good"]
performance_order = ["Below Avg", "Average", "Good", "Excellent"]
```

Step 2: Encode

```python
oe = OrdinalEncoder(
    categories=[study_order, support_order, performance_order]
)

df[["study_hours_ord",
    "parent_support_ord",
    "performance_ord"]] = oe.fit_transform(
    df[["study_hours", "parent_support", "performance"]]
)

df
```


##Dataset2


```python


data = {
    "user": ["Rohit", "Ananya", "Suresh", "Meena", "Kabir"],
    "difficulty": ["Easy", "Medium", "Hard", "Medium", "Easy"],
    "content_quality": ["Poor", "Good", "Excellent", "Average", "Good"],
    "satisfaction": ["Low", "Medium", "High", "High", "Medium"]
}

df = pd.DataFrame(data)
df
```

```python
#Define order
diff_order=["Easy","Medium","Hard"]
quality_order=["Poor","Average","Good","Excellent"]
satisfaction_order=["Low","Medium","High"]
```

```python
from sklearn.preprocessing import OrdinalEncoder

oe = OrdinalEncoder(
    categories=[diff_order, quality_order, satisfaction_order]
)

df[[
    "difficulty_ord",
    "content_quality_ord",
    "satisfaction_ord"
]] = oe.fit_transform(
    df[["difficulty", "content_quality", "satisfaction"]]
)

df
```

#Dataset 3 Label Encoder for catgorical output

```python
import pandas as pd

data = {
    "user": ["Rohit", "Ananya", "Suresh", "Meena", "Kabir"],
    "course": ["Python", "Java", "Python", "C++", "Java"],
    "city": ["Delhi", "Mumbai", "Delhi", "Pune", "Mumbai"]
}

df = pd.DataFrame(data)
df
```

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

df["course_le"] = le.fit_transform(df["course"])
df["city_le"] = le.fit_transform(df["city"])

df
```
